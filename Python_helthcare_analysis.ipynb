```python
##  Histogram by age

plt.figure(figsize=(9, 6))
df.Age.hist(edgecolor='Green', grid=False)
plt.xlabel('Age', fontsize=12)
plt.ylabel('patients Count', fontsize=12)
plt.title('Age Distribution of Patients ', fontsize=14, fontweight='bold')
plt.show()
```

```python
## What is the status of patient based on the test results in percentages.

df['Test Results'].value_counts().plot(figsize=(6, 6), kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Status of patient based on the test result', fontweight='bold')
plt.ylabel('')
plt.show()
```

```python
## What is the percentage of each gender.

df['Gender'].value_counts(normalize = True) *  100
```

```python
## How many blood type are availbale

df['Blood Type'].value_counts()
```

```python
## What is the gender distribution of blood types?

df.groupby('Gender')['Blood Type'].value_counts()
```

```python
## What is the distribution of medical conditions ?

df['Medical Condition'].value_counts()
```

```python
## Which gender is more affected by the disease?

df.groupby('Medical Condition')['Gender'].value_counts(normalize = True) * 100
```

```python
## What is the number of patients broken down by sex, medical condition, and blood type?

Group_func = df.groupby(['Medical Condition', 'Gender', 'Blood Type']).count()['Name']
sorted = Group_func.unstack().sort_values( by = ['Medical Condition'], ascending=False)
sorted
```

```python
## distribute Medication
df['Medication'].value_counts()
```

```python
## What is the distribution of medical conditions for each type of medication?

df.groupby('Medication')['Medical Condition'].value_counts()
```

```python
## Group by Gender Admission Types
df.groupby('Gender')['Admission Type'].value_counts().unstack()
```

```python
##Group by Admission Types Mediaction
df.groupby('Admission Type')['Medication'].value_counts()
```

```python
##What medications are most commonly used for each medical condition within each type of admission?

df.groupby(["Admission Type", "Medication", "Medical Condition"]).count()["Name"].unstack()
```
